# Simple multilayer neural network

**3 inputs → 2 hidden neurons → 1 output neuron**  
One training example and one weight update, following the first example in `lab3.ipynb`.


## 1. Task


In the previous example, `w` contained the weights of the only layer (3 inputs → 1 output neuron). Here there are two layers, so each has its own weights: `w1` connects the inputs to the hidden neurons, and `w2` connects the hidden neurons to the output neuron. The same numbering applies to `b1` and `b2`.


```python
import numpy as np

x = np.array([0.5, -0.2, 0.15])

# w1: weights from the 3 inputs to the 2 hidden neurons. Each column belongs to one neuron.
w1 = np.array([
    [0.2,  0.4],
    [2.0, -0.5],
    [0.1,  0.3]
])
b1 = np.array([1.0, 0.2])

# w2: weights from the 2 hidden neurons to the output neuron.
w2 = np.array([0.3, -0.4])
b2 = 0.1

y_true = 1
lr = 0.1

def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

## 2. Forward pass


```python
hidden_net = np.dot(x, w1) + b1
hidden_output = sigmoid(hidden_net)

net = np.dot(hidden_output, w2) + b2
y_pred = sigmoid(net)

loss = 0.5 * (y_true - y_pred) ** 2

print("Hidden layer:", hidden_output)
print("Prediction:", y_pred)
print("Loss:", loss)
```

## 3. Backward pass


```python
# Output neuron error: the same calculation as in the single-neuron example.
output_error = (y_pred - y_true) * (1 - y_pred) * y_pred

# Pass the error back to both hidden neurons, using the original w2.
hidden_error = output_error * w2 * (1 - hidden_output) * hidden_output

print("Output error:", output_error)
print("Hidden errors:", hidden_error)
```

## 4. Weight update


```python
# Update the weights and bias of the output neuron.
w2 = w2 - lr * output_error * hidden_output
b2 = b2 - lr * output_error

# Update the weights and biases of both hidden neurons.
# The outer product produces a 3 × 2 matrix with the same shape as w1.
w1 = w1 - lr * np.outer(x, hidden_error)
b1 = b1 - lr * hidden_error

print("w1:\n", w1)
print("b1:", b1)
print("w2:", w2)
print("b2:", b2)
```
